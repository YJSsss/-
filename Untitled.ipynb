```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import xgboost

from scipy import stats
from imblearn.over_sampling import SMOTE
from sklearn import model_selection
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
from sklearn import tree
from sklearn import metrics
from sklearn import ensemble
from sklearn import svm
from sklearn.neural_network import MLPClassifier
```

```python
vbox = pd.read_csv(r'VBOX_standby.csv', sep = ',')
```

```python
vbox.head()
```

```python
vbox.shape
```

```python
vbox.dtypes
```

```python
# 加速度单位换算，乘g
vbox.Latacc = vbox.Latacc*9.8
vbox.Longacc = vbox.Longacc*9.8
# 数据类型转换
vbox.sats = vbox.sats.astype('float64')
```

```python
# 删除卫星数小于3或经纬度为0的数据
vbox.drop(vbox[(vbox.sats <= 3)|(vbox.long == 0)|(vbox.lat == 0)].index, inplace = True)
# - 识别不了（不知道为啥，所以改成 _
vbox.rename(columns={'vert-vel':'vert_vel'}, inplace = True)
# 删除下面几项同时为0的数据
vbox.drop(vbox[(vbox.velocity == 0)&(vbox.heading == 0)&(vbox.vert_vel == 0)&(vbox.Longacc	== 0)&(vbox.Latacc == 0)].index, inplace = True)
```

```python
# 删除无意义变量，此处指sats、time、long、lat、height、vert-vel、GPS-Longacc、rel-height
vbox.drop(['sats', 'time', 'long', 'lat', 'height', 'vert_vel', 'GPS-Longacc', 'rel-height'], axis = 1, inplace = True)
# 检查删除后的数据集中各变量缺失值情况
vbox.isnull().sum(axis = 0)
```

```python
# 统计异常驾驶与正常驾驶的分布
vbox['event'].value_counts()
```

```python
# 添加指标，利用rolling-windows，添加均值、方差
# 每五个计算一次，最小计算周期为1，缺失值用后一个值填补
vbox['velocity_mean'] = vbox['velocity'].rolling(5, min_periods = 1).mean()
vbox['velocity_std'] = vbox['velocity'].rolling(5, min_periods = 1).std().fillna(method='bfill')
vbox['heading_std'] = vbox['heading'].rolling(5, min_periods = 1).std().fillna(method='bfill')
vbox['turn-radius_mean'] = vbox['turn-radius'].rolling(5, min_periods = 1).mean()
vbox['turn-radius_std'] = vbox['turn-radius'].rolling(5, min_periods = 1).std().fillna(method='bfill')
vbox['Centre-Line-Deviation_std'] = vbox['Centre-Line-Deviation'].rolling(5, min_periods = 1).std().fillna(method='bfill')
vbox['AbsHead_std'] = vbox['AbsHead'].rolling(5, min_periods = 1).std().fillna(method='bfill')
vbox['Longacc_mean'] = vbox['Longacc'].rolling(5, min_periods = 1).mean()
vbox['Longacc_std'] = vbox['Longacc'].rolling(5, min_periods = 1).std().fillna(method='bfill')
vbox['Latacc_mean'] = vbox['Latacc'].rolling(5, min_periods = 1).mean()
vbox['Latacc_std'] = vbox['Latacc'].rolling(5, min_periods = 1).std().fillna(method='bfill')

vbox.head(549)
```

```python
# 相关性分析
# vboxcorr = vbox.drop('event', axis = 1, inplace = False).corr(method = 'pearson')
vboxcorr = vbox.corr(method = 'pearson')
vboxcorr.head()
```

```python
# 相关系数热力图
plt.figure(figsize=(15, 13), dpi = 500)
corr_hm = sns.heatmap(data = vboxcorr, cmap = 'YlGnBu', square = True, annot=True)
# cmap：配色
# square：每个子图是否以正方形显示
# annot：子图内数值是否显示

plt.savefig('corr_hm.png', dpi = 500)  
plt.show()
```

```python
# 统计异常驾驶与正常驾驶的分布
counts = vbox['event'].value_counts()
print(counts)
# 美化饼图
explode = [0, 0.3]  # 突出显示abnormal
colors = ['#7AE5C9', '#E59EEB']
# 绘制饼图看数据集是否处于不平衡状态
plt.pie(x = counts, # 绘图数据
        explode = explode, 
        labels = pd.Series(counts.index).map({0:'normal',1:'abnormal'}), # 添加文字标签
        colors = colors, 
        autopct = '%.2f%%',  # 设置百分比的格式，这里保留两位小数
        wedgeprops = {'linewidth':1.3, 'edgecolor':'black'}
       )

plt.savefig('ab_distribution.png', dpi = 500) 
# 显示图形
plt.show()
```

```python
vbox.drop(['Longacc', 'Latacc', 'velocity_mean', 'turn-radius_mean', 'turn-radius', 'Centre-Line-Deviation_std', 'AbsHead_std', 'Longacc_std', 'Latacc_std'], 
          inplace = True, axis = 1)
```

```python
# 将数据拆分为训练集和测试集
X = vbox.drop(['event'], axis = 1)
y = vbox.event
# 数据拆分
X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size = 0.3, random_state = 1234)
```

```python
# 运用SMOTE算法重采样实现训练数据集的平衡
over_samples = SMOTE(random_state = 1234)
over_samples_X, over_samples_y = over_samples.fit_resample(X_train, y_train)
# 重抽样后的类别比例
print(pd.Series(over_samples_y).value_counts()/len(over_samples_y))
```

```python
over_samples_y.shape
```

```python
# 对重采样训练集数据及测试集数据进行标准化处理
ss = StandardScaler()
over_samples_X = ss.fit_transform(over_samples_X)
X_test = ss.transform(X_test)
```

## 决策树模型

```python
# 预设各参数的不同选项值
max_depth = [10, 11, 12, 13, 14, 15, 16, 17, 18]  #决策树的最大深度
min_samples_split = [2, 4, 6, 8]  #根节点或中间节点能够继续分割的最小样本量
min_samples_leaf = [2, 4, 8, 10, 12]  #叶节点的最小样本量
# 将各参数值以字典形式组织起来
parameters = {'max_depth':max_depth, 'min_samples_split':min_samples_split, 'min_samples_leaf':min_samples_leaf}

# 网格搜索法，测试不同的参数值
grid_dtcateg = GridSearchCV(estimator = tree.DecisionTreeClassifier(), param_grid = parameters, cv = 10)
# 模型拟合
grid_dtcateg.fit(over_samples_X, over_samples_y)
# 返回最佳组合的参数值
grid_dtcateg.best_params_
```

```python
# 构建分类决策树
CART_Class = tree.DecisionTreeClassifier(max_depth = 18, min_samples_leaf = 2, min_samples_split = 4)
# 模型拟合
decision_tree = CART_Class.fit(over_samples_X, over_samples_y)
# 模型用测试集进行预测
pred = CART_Class.predict(X_test)

# 检验模型准确性
print('模型在训练集的预测准确性：\n', metrics.accuracy_score(over_samples_y, CART_Class.predict(over_samples_X)))
print('模型在测试集的预测准确性：\n', metrics.accuracy_score(y_test, pred))
print('模型的评估报告：\n', metrics.classification_report(y_test, pred))
```

```python
# 绘制ROC曲线
y_score = CART_Class.predict_proba(X_test)[:, 1]
fpr, tpr, threshold = metrics.roc_curve(y_test, y_score)
# 计算AUC值
roc_auc = metrics.auc(fpr, tpr)
# 绘制面积图
plt.stackplot(fpr, tpr, color = 'grey', alpha = 0.5, edgecolor = 'black')
# 绘制边际线
plt.plot(fpr, tpr, color = 'black', lw = 1)
# 绘制对角线
plt.plot([0, 1], [0, 1], color = 'blue', linestyle = '--')
# 添加文本信息
plt.text(0.5, 0.3, 'ROC curve (area = %0.4f)'% roc_auc)
plt.xlabel('1-Specificity')
plt.ylabel('Sensitivity')

plt.savefig('DT_ROC.png', dpi = 500)  
plt.show()
```

```python
# 混淆矩阵
cm = metrics.confusion_matrix(y_test, pred)

Accuracy = metrics.accuracy_score(y_test, pred)
Sensitivity = metrics.recall_score(y_test, pred)
Specificity = metrics.recall_score(y_test, pred, pos_label=0)
print('模型准确率为%.2f%%' %(Accuracy*100))
print('正例覆盖率为%.2f%%' %(Sensitivity*100))
print('负例覆盖率为%.2f%%' %(Specificity*100))

# 绘制热力图
sns.heatmap(cm, annot = True, fmt = 'g', cmap = 'YlGnBu')
# 添加横纵坐标文本信息
plt.xlabel('Predicted Label')
plt.ylabel('True Label')

plt.savefig('DT_CM.png', dpi = 500)  
plt.show()
```

## XGBoost模型

```python
# 构建XGBoost分类器
xgboost = xgboost.XGBClassifier(learning_rate = 0.3, 
                                subsample = 0.9,
                                max_depth = 7, 
                                colsample_bytree = 0.8, 
                                eval_metric=['logloss','auc','error'], use_label_encoder = False)
# subsample：默认值1，取值在(0, 1)之间，设置为1表示使用所有数据训练弱学习器。如果取值小于1，则只有一部分样本会去做GBDT的决策树拟合。
# 选择小于1的比例可以减少方差，即防止过拟合，但是会增加样本拟合的偏差，因此取值不能太低
# colsample_bytree：构建弱学习器时，对特征随机采样的比例，默认值为1。推荐的候选值为：[0.6, 0.7, 0.8, 0.9, 1]

# 使用重抽样后的数据，对其建模
xgboost.fit(over_samples_X,over_samples_y)
# 将模型运用到测试数据集中
resample_pred = xgboost.predict(np.array(X_test))

# 返回模型的预测效果
print('模型的准确率为：\n', metrics.accuracy_score(y_test, resample_pred))
print('模型的评估报告：\n', metrics.classification_report(y_test, resample_pred))
```

```python
# 绘制ROC曲线
y_score = xgboost.predict_proba(np.array(X_test))[:,1]
fpr, tpr, threshold = metrics.roc_curve(y_test, y_score)
# 计算AUC值
roc_auc = metrics.auc(fpr, tpr)

# 绘制面积图
plt.stackplot(fpr, tpr, color = 'grey', alpha = 0.5, edgecolor = 'black')
# 绘制边际线
plt.plot(fpr, tpr, color = 'black', lw = 1)
# 添加对角线
plt.plot([0, 1], [0, 1], color = 'blue', linestyle = '--')
# 添加文本信息
plt.text(0.5, 0.3, 'ROC curve (area = %0.4f)' % roc_auc)
# 添加x轴与y轴标签
plt.xlabel('1-Specificity')
plt.ylabel('Sensitivity')

plt.savefig('XGB_ROC.png', dpi = 500) 
plt.show()
```

```python
# 混淆矩阵
cm = metrics.confusion_matrix(y_test, resample_pred)

Accuracy = metrics.accuracy_score(y_test, resample_pred)
Sensitivity = metrics.recall_score(y_test, resample_pred)
Specificity = metrics.recall_score(y_test, resample_pred, pos_label=0)
print('模型准确率为%.2f%%' %(Accuracy*100))
print('正例覆盖率为%.2f%%' %(Sensitivity*100))
print('负例覆盖率为%.2f%%' %(Specificity*100))

# 绘制热力图
sns.heatmap(cm, annot = True, fmt = 'g', cmap = 'YlGnBu')
# 添加横纵坐标文本信息
plt.xlabel('Predicted Label')
plt.ylabel('True Label')

plt.savefig('XGB_CM.png', dpi = 500) 
plt.show()
```
